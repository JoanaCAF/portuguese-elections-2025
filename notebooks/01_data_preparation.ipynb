# 01 — Data Preparation: Building the Main Feature Table
Merges Censos 2021 sociodemographic data with 2021/2022 municipal and legislative election results into a single feature table used for modelling.


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from functools import reduce
```

```python
escolaridade_por_concelho = pd.read_csv('censos_2021_escolaridade_por_concelho.csv', sep=';')
estado_civil = pd.read_csv('censos_2021_estado_civil.csv', sep=';')
taxa_imigrantes = pd.read_csv('censos_2021_taxa_imigrantes_concelho.csv', sep=';')
local_residencia = pd.read_csv('censos_2021_local_residencia.csv', sep=';')
poder_compra = pd.read_csv('censos_2021_poder_de_compra.csv', sep=';')
populacao_concelho = pd.read_csv('censos_2021_populacao_por_concelho.csv', sep=';')
taxa_criminalidade = pd.read_csv('censos_2021_taxa_criminalidade.csv', sep=';')
partido_vencedor_CM_2022 = pd.read_csv('resultado_partido_vencedor_cm_2022.csv', sep=';')
mandados_2025 = pd.read_csv('Concelhos_Autarquicas_2025_v0.csv', sep=',', encoding='latin1')
resultados_legislativas_2022 = pd.read_csv('Legislativas_2022_por_Freguesia_limpo.csv', sep=';')
resultados_autarquicas_2021 = pd.read_csv('resultados_CM_2021.csv', sep=';')
```

```python
lista_dfs = [
    escolaridade_por_concelho,
    estado_civil,
    taxa_imigrantes,
    local_residencia,
    poder_compra,
    populacao_concelho,
    taxa_criminalidade,
    partido_vencedor_CM_2022,
    mandados_2025,
    resultados_legislativas_2022,
    resultados_autarquicas_2021
]
```

```python
def verificar_primeira_coluna(lista_dfs):
    for i, df in enumerate(lista_dfs):
        primeira_coluna = df.columns[0]
        print(f"DataFrame {i+1}: primeira coluna = {primeira_coluna}")

verificar_primeira_coluna(lista_dfs)
```

```python
poder_compra.rename(columns={"Concelho_Limpo": "Concelho"}, inplace=True)
partido_vencedor_CM_2022.rename(columns={"CONC":"Concelho"},inplace=True)
mandados_2025.rename(columns={"CONC":"Concelho"},inplace=True)
resultados_autarquicas_2021.rename(columns={"CONC":"Concelho"},inplace=True)
```

```python
resultados_legislativas_2022 = resultados_legislativas_2022.rename(columns=lambda x: f"{x}_2022" if x != "Concelho" else x)
```

```python
verificar_primeira_coluna(lista_dfs)
```

```python
for i, df in enumerate(lista_dfs):
    for col in df.columns:
        if df[col].dtype == "object":  # só mexe em colunas de texto
            df[col] = df[col].str.replace(",", ".", regex=False)
            df[col] = pd.to_numeric(df[col], errors="ignore")
    lista_dfs[i] = df  # atualiza o DataFrame na lista

```

```python
mandados_2025['Limite'] = mandados_2025['Limite'].map({'Não': 0, 'Sim': 1})
```

```python
partido_vencedor_CM_2022 = partido_vencedor_CM_2022.drop(columns=['cor', 'Percentagem_Max','CONC_std'])
partido_vencedor_CM_2022 = pd.get_dummies(partido_vencedor_CM_2022, columns=['Partido_Vencedor'], dtype =int)
```

```python
resultados_legislativas_2022 = resultados_legislativas_2022.fillna(0)
```

```python
lista_final_dfs = [escolaridade_por_concelho,
    estado_civil,
    taxa_imigrantes,
    local_residencia,
    poder_compra,
    populacao_concelho,
    taxa_criminalidade,
    partido_vencedor_CM_2022,
    mandados_2025,
    resultados_legislativas_2022,
    resultados_autarquicas_2021]
```

```python
tabela_principal = reduce(lambda left, right: pd.merge(left, right, on='Concelho', how='outer'), lista_final_dfs)
```
